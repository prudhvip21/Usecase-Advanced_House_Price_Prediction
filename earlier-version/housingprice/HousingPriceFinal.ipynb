```python
# Statistical EDA

# Test of Independence

#Analysis of Categorical variables
#Chi square tests

from scipy.stats import chi2_contingency
factors_paired=[(i,j) for i in cat1.columns for j in cat1.columns]
chi2, p_values = [],[]
for f in factors_paired:
    if f[0]!=f[1]:
        chitest=chi2_contingency(pd.crosstab(price[f[0]],price[f[1]]))
        chi2.append(chitest[0])
        p_values.append(chitest[1])
    else:
        chi2.append(0)
        p_values.append(0)


p1=np.array(p_values).reshape((60,60))
#p1=np.array(p_values)
p1=pd.DataFrame(p1.round(2), index=cat1.columns, columns=cat1.columns)
p1
```

```python
# Null Hypothesis: Independant:(p value>0.05) # Accept Null Hypothesis 

# Alternate Hypothesis: Dependant:(p is <0.05) # Accept Alternate Hypothesis
p2= p1[p1<0.05]

#Nan=Independant

#Getting Dependant
for i in p2.index:
    for j in p2.columns:
        if ~np.isnan(p2.loc[i,j]) and i!=j:
            print((i,j))
```

```python
#Model Building
price2=price.copy()

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
price2.info()
#Transforming Categorical to Discrete
for col in price2.select_dtypes('category'):
    price2[col] = le.fit_transform(price2[col].astype(str) )
price2.head()
```

```python
import statsmodels.formula.api as sm
from sklearn import linear_model
x=price2.drop(['SalePrice'],axis=1)
y=price2['SalePrice']
model = sm.ols(formula='y~x',data=price2)
fitted=model.fit()
fitted.summary()
```

```python
new=price2[price2.columns[fitted.pvalues <= 0.05]]
x1=new.drop(['SalePrice'],axis=1)
y1=new['SalePrice']

model1 = sm.ols(formula='y1~x1',data=new)
fitted1=model1.fit()
fitted1.summary()
```

```python
#Selecting Important variables using Variable Selection Method
#Variable Selections using Recursive Feature Elimination
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.feature_selection import RFE
# price2.replace(['inf','-inf'], np.nan)
# price2.dropna(inplace=True)

x=price2.drop(['SalePrice'],axis=1)
y=price2['SalePrice']
#Recursive feature Elimination
model = LinearRegression()
#Initializing RFE model
rfe = RFE(model,80)
#Transforming data using RFE
X_rfe = rfe.fit_transform(x,y)  
#Fitting the data to model
model.fit(X_rfe,y)
# print(rfe.support_)
# print(rfe.ranking_)
```

```python
#no of features
from sklearn.model_selection import train_test_split
nof_list=np.arange(1,80)            
high_score=0
#Variable to store the optimum features
nof=0           
score_list =[]
for n in range(len(nof_list)):
    X_train, X_test, y_train, y_test = train_test_split(x,y, test_size = 0.3, random_state = 0)
    model = LinearRegression()
    rfe = RFE(model,nof_list[n])
    X_train_rfe = rfe.fit_transform(X_train,y_train)
    X_test_rfe = rfe.transform(X_test)
    model.fit(X_train_rfe,y_train)
    score = model.score(X_test_rfe,y_test)
    score_list.append(score)
    if(score>high_score):
        high_score = score
        nof = nof_list[n]
print("Optimum number of features: %d" %nof)
print("Score with %d features: %f" % (nof, high_score))
```

```python
#Plotting graph for feature Selection
plt.figure(figsize=(19,8))
sns.lineplot(x=nof_list, y=score_list)
plt.xlabel('No of Features', fontsize=14)
plt.ylabel("ScoreList",fontsize=14)
plt.xticks(rotation=20,fontsize=10)
plt.xticks(rotation=20,fontsize=10)
plt.title("[Features vs ScoreList]",fontsize=16).set_weight('bold')
plt.show()
```
